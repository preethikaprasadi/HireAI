```python
def generate_experience(job_title, skills, total_years):
    experience = []

    remaining_years = total_years
    current_year = 2026

    number_of_jobs = 1

    if total_years >= 4:
        number_of_jobs = random.choice([2, 2, 3])

    for index in range(number_of_jobs):

        if index == number_of_jobs - 1:
            years_at_job = max(1, remaining_years)
        else:
            max_years = max(1, remaining_years - (number_of_jobs - index - 1))
            years_at_job = random.randint(1, min(3, max_years))

        start_year = current_year - years_at_job
        end_year = current_year

        company = random.choice(COMPANIES)

        job = {
            "company": company,
            "job_title": job_title,
            "start_date": f"{start_year}-01",
            "end_date": None if index == 0 else f"{end_year}-01",
            "description": (
                f"Worked as a {job_title} at {company}, "
                f"developing and maintaining software solutions "
                f"using {', '.join(skills[:3])}."
            )
        }

        experience.append(job)

        remaining_years -= years_at_job
        current_year = start_year

        if remaining_years <= 0:
            break

    return experience
```

```python
def generate_candidate():
    first_name = random.choice(FIRST_NAMES)
    last_name = random.choice(LAST_NAMES)

    name = f"{first_name} {last_name}"
    email = f"{first_name.lower()}.{last_name.lower()}@example.com"

    phone = (
        f"+94 7{random.randint(0, 9)} "
        f"{random.randint(100, 999)} "
        f"{random.randint(1000, 9999)}"
    )

    job_title = random.choice(list(ROLE_PROFILES.keys()))

    available_skills = ROLE_PROFILES[job_title]["skills"]

    number_of_skills = random.randint(
        min(4, len(available_skills)),
        min(7, len(available_skills))
    )

    skills = random.sample(
        available_skills,
        number_of_skills
    )

    years = random.randint(1, 7)

    experience = generate_experience(
        job_title=job_title,
        skills=skills,
        total_years=years
    )

    degree = random.choice(DEGREES)
    university = random.choice(UNIVERSITIES)

    start_year = 2026 - years

    graduation_year = random.randint(
        max(2018, start_year - 4),
        max(2019, start_year - 1)
    )

    certification = random.choice(CERTIFICATIONS)

    project_name = random.choice(PROJECT_NAMES)

    project_template = random.choice(
        PROJECT_TEMPLATES[job_title]
    )

    project_description = project_template.format(
        skills=", ".join(skills[:3])
    )

    projects = [
        {
            "name": project_name,
            "description": project_description
        }
    ]

    candidate = {
        "candidate": {
            "name": name,
            "email": email,
            "phone": phone
        },

        "summary": (
            f"{job_title} with {years} years of professional "
            f"experience specializing in "
            f"{', '.join(skills[:3])}."
        ),

        "skills": skills,

        "years_of_experience": years,

        "job_titles": [job_title],

        "experience": experience,

        "education": [
            {
                "degree": degree,
                "institution": university,
                "year": graduation_year
            }
        ],

        "projects": projects,

        "certifications": [certification]
    }

    return candidate
```

```python
def format_header(candidate):
    info = candidate["candidate"]

    return (
        f"{info['name']}\n"
        f"Email: {info['email']}\n"
        f"Phone: {info['phone']}"
    )
```

```python
def resume_style_traditional(candidate):
    sections = []

    sections.append(format_header(candidate))

    sections.append(
        f"{random.choice(SECTION_TITLES['summary'])}\n"
        f"{candidate['summary']}"
    )

    sections.append(
        f"{random.choice(SECTION_TITLES['skills'])}\n"
        + ", ".join(candidate["skills"])
    )

    experience_text = []

    for exp in candidate["experience"]:
        end_date = exp["end_date"] or "Present"

        experience_text.append(
            f"{exp['job_title']}\n"
            f"{exp['company']}\n"
            f"{exp['start_date']} - {end_date}\n"
            f"{exp['description']}"
        )

    sections.append(
        f"{random.choice(SECTION_TITLES['experience'])}\n"
        + "\n\n".join(experience_text)
    )

    education_text = []

    for edu in candidate["education"]:
        education_text.append(
            f"{edu['degree']}, "
            f"{edu['institution']}, "
            f"{edu['year']}"
        )

    sections.append(
        f"{random.choice(SECTION_TITLES['education'])}\n"
        + "\n".join(education_text)
    )

    project_text = []

    for project in candidate["projects"]:
        project_text.append(
            f"{project['name']}\n"
            f"{project['description']}"
        )

    sections.append(
        f"{random.choice(SECTION_TITLES['projects'])}\n"
        + "\n\n".join(project_text)
    )

    sections.append(
        f"{random.choice(SECTION_TITLES['certifications'])}\n"
        + "\n".join(candidate["certifications"])
    )

    return "\n\n".join(sections)
```

```python
def resume_style_compact(candidate):
    info = candidate["candidate"]

    text = (
        f"{info['name']} | "
        f"{info['email']} | "
        f"{info['phone']}\n\n"
    )

    text += (
        f"{candidate['summary']}\n\n"
    )

    text += (
        f"Skills: "
        f"{', '.join(candidate['skills'])}\n\n"
    )

    text += "Experience:\n"

    for exp in candidate["experience"]:
        end_date = exp["end_date"] or "Present"

        text += (
            f"{exp['job_title']} at {exp['company']} "
            f"({exp['start_date']} - {end_date})\n"
            f"{exp['description']}\n\n"
        )

    text += "Education:\n"

    for edu in candidate["education"]:
        text += (
            f"{edu['degree']} - "
            f"{edu['institution']} "
            f"({edu['year']})\n"
        )

    text += "\nProjects:\n"

    for project in candidate["projects"]:
        text += (
            f"{project['name']}: "
            f"{project['description']}\n"
        )

    text += "\nCertifications:\n"

    for certification in candidate["certifications"]:
        text += f"- {certification}\n"

    return text
```

```python
def resume_style_narrative(candidate):
    info = candidate["candidate"]

    text = (
        f"{info['name']}\n"
        f"{info['email']} | {info['phone']}\n\n"
    )

    text += (
        f"Professional Profile\n"
        f"{candidate['summary']}\n\n"
    )

    text += (
        "Technical background includes "
        f"{', '.join(candidate['skills'])}.\n\n"
    )

    text += "Career History\n"

    for exp in candidate["experience"]:
        end_date = exp["end_date"] or "Present"

        text += (
            f"{exp['job_title']} at {exp['company']} "
            f"from {exp['start_date']} to {end_date}. "
            f"{exp['description']}\n\n"
        )

    text += "Academic Background\n"

    for edu in candidate["education"]:
        text += (
            f"{edu['degree']} from "
            f"{edu['institution']} in "
            f"{edu['year']}.\n"
        )

    text += "\nSelected Projects\n"

    for project in candidate["projects"]:
        text += (
            f"{project['name']} - "
            f"{project['description']}\n"
        )

    text += "\nProfessional Certifications\n"

    for certification in candidate["certifications"]:
        text += f"{certification}\n"

    return text
```

```python
def add_formatting_variation(text):
    variation = random.choice([
        "normal",
        "extra_spaces",
        "uppercase_headers",
        "bullet_style"
    ])

    if variation == "extra_spaces":
        text = re.sub(r"\n\n+", "\n\n\n", text)

    elif variation == "uppercase_headers":
        lines = text.splitlines()

        new_lines = []

        for line in lines:
            if (
                line.strip().upper() in [
                    "SUMMARY",
                    "PROFESSIONAL SUMMARY",
                    "PROFILE",
                    "ABOUT ME",
                    "SKILLS",
                    "TECHNICAL SKILLS",
                    "CORE SKILLS",
                    "TECHNOLOGIES",
                    "EXPERIENCE",
                    "WORK EXPERIENCE",
                    "PROFESSIONAL EXPERIENCE",
                    "EMPLOYMENT HISTORY",
                    "EDUCATION",
                    "PROJECTS",
                    "PERSONAL PROJECTS",
                    "KEY PROJECTS",
                    "CERTIFICATIONS",
                    "CERTIFICATES"
                ]
            ):
                new_lines.append(line.upper())
            else:
                new_lines.append(line)

        text = "\n".join(new_lines)

    elif variation == "bullet_style":
        lines = text.splitlines()

        new_lines = []

        for line in lines:
            if line.startswith("Worked as"):
                line = "- " + line

            new_lines.append(line)

        text = "\n".join(new_lines)

    return text
```

```python
def generate_resume_text(candidate):
    style = random.choice([
        "traditional",
        "compact",
        "narrative"
    ])

    if style == "traditional":
        text = resume_style_traditional(candidate)

    elif style == "compact":
        text = resume_style_compact(candidate)

    else:
        text = resume_style_narrative(candidate)

    text = add_formatting_variation(text)

    return text
```

```python
def validate_candidate(candidate):
    required_top_level_fields = [
        "candidate",
        "summary",
        "skills",
        "years_of_experience",
        "job_titles",
        "experience",
        "education",
        "projects",
        "certifications"
    ]

    for field in required_top_level_fields:
        if field not in candidate:
            return False, f"Missing field: {field}"

    candidate_fields = [
        "name",
        "email",
        "phone"
    ]

    for field in candidate_fields:
        if field not in candidate["candidate"]:
            return False, f"Missing candidate field: {field}"

    if not isinstance(candidate["skills"], list):
        return False, "skills must be a list"

    if not isinstance(candidate["experience"], list):
        return False, "experience must be a list"

    if not isinstance(candidate["education"], list):
        return False, "education must be a list"

    if not isinstance(candidate["projects"], list):
        return False, "projects must be a list"

    if not isinstance(candidate["certifications"], list):
        return False, "certifications must be a list"

    if not isinstance(
        candidate["years_of_experience"],
        int
    ):
        return False, "years_of_experience must be an integer"

    return True, "Valid"
```

```python
candidate = generate_candidate()

valid, message = validate_candidate(candidate)

print("Valid:", valid)
print("Message:", message)
```

```python
def generate_training_example():
    candidate = generate_candidate()

    resume_text = generate_resume_text(candidate)

    valid, message = validate_candidate(candidate)

    if not valid:
        raise ValueError(message)

    return {
        "instruction": (
            "Extract structured candidate information "
            "from the resume."
        ),
        "input": resume_text,
        "output": candidate
    }
```

```python
example = generate_training_example()

print("INSTRUCTION:")
print(example["instruction"])

print("\nINPUT:")
print(example["input"])

print("\nOUTPUT:")
print(json.dumps(example["output"], indent=2))
```

```python
DATASET_SIZE = 500

dataset = []

for i in range(DATASET_SIZE):
    example = generate_training_example()
    dataset.append(example)

print("Generated examples:", len(dataset))
```

```python
resume_texts = [
    example["input"].strip().lower()
    for example in dataset
]

unique_resume_texts = set(resume_texts)

print("Total examples:", len(dataset))
print("Unique resumes:", len(unique_resume_texts))
print("Duplicates:", len(dataset) - len(unique_resume_texts))
```

```python
unique_dataset = []
seen = set()

for example in dataset:
    key = example["input"].strip().lower()

    if key not in seen:
        seen.add(key)
        unique_dataset.append(example)

dataset = unique_dataset

print("Final dataset size:", len(dataset))
```

```python
role_counter = Counter()

experience_counter = Counter()

skill_counter = Counter()

for example in dataset:
    output = example["output"]

    for job_title in output["job_titles"]:
        role_counter[job_title] += 1

    experience_counter[
        output["years_of_experience"]
    ] += 1

    for skill in output["skills"]:
        skill_counter[skill] += 1

print("JOB TITLE DISTRIBUTION")
print("----------------------")

for role, count in role_counter.most_common():
    print(f"{role}: {count}")

print("\nEXPERIENCE DISTRIBUTION")
print("-----------------------")

for years, count in sorted(experience_counter.items()):
    print(f"{years} years: {count}")

print("\nTOP SKILLS")
print("----------")

for skill, count in skill_counter.most_common(15):
    print(f"{skill}: {count}")
```

```python
def load_jsonl(path):
    records = []

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:
            records.append(
                json.loads(line)
            )

    return records
```
