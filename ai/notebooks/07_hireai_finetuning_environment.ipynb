```python
import sys
import platform
import psutil
import shutil
from pathlib import Path

# ==============================
# 1. Python & System Information
# ==============================

print("=" * 60)
print("SYSTEM INFORMATION")
print("=" * 60)

print("Python:", sys.version)
print("Platform:", platform.platform())
print("Processor:", platform.processor())


# ==============================
# 2. RAM Information
# ==============================

memory = psutil.virtual_memory()

print("\nRAM INFORMATION")
print("-" * 60)

print(f"Total RAM:     {memory.total / (1024**3):.2f} GB")
print(f"Available RAM: {memory.available / (1024**3):.2f} GB")
print(f"Used RAM:      {memory.used / (1024**3):.2f} GB")
print(f"RAM usage:     {memory.percent}%")


# ==============================
# 3. GPU / PyTorch Information
# ==============================

print("\nPYTORCH / GPU INFORMATION")
print("-" * 60)

try:
    import torch

    print("PyTorch:", torch.__version__)
    print("CUDA available:", torch.cuda.is_available())

    if torch.cuda.is_available():
        print("GPU:", torch.cuda.get_device_name(0))
        print("CUDA version:", torch.version.cuda)

        gpu_memory = torch.cuda.get_device_properties(0).total_memory

        print(
            f"GPU VRAM: {gpu_memory / (1024**3):.2f} GB"
        )
    else:
        print("Using CPU")

except ImportError:
    print("PyTorch is not installed.")


# ==============================
# 4. Disk Information
# ==============================

# Change this to your actual project location.
# Example:
# PROJECT_ROOT = r"E:\MyProject"

PROJECT_ROOT = Path.cwd()

total, used, free = shutil.disk_usage(PROJECT_ROOT)

print("\nDISK INFORMATION")
print("-" * 60)

print("Checking disk:", PROJECT_ROOT)

print(f"Total: {total / (1024**3):.2f} GB")
print(f"Used:  {used / (1024**3):.2f} GB")
print(f"Free:  {free / (1024**3):.2f} GB")
```

```python
from pathlib import Path

V3_TRAIN = (
    PROJECT_ROOT
    / "ai"
    / "datasets"
    / "processed"
    / "hireai_train_v3.jsonl"
)

V3_VALIDATION = (
    PROJECT_ROOT
    / "ai"
    / "datasets"
    / "processed"
    / "hireai_validation_v3.jsonl"
)

V3_TEST = (
    PROJECT_ROOT
    / "ai"
    / "datasets"
    / "processed"
    / "hireai_test_v3.jsonl"
)

for path in [
    V3_TRAIN,
    V3_VALIDATION,
    V3_TEST
]:

    print(
        path.name,
        "exists:",
        path.exists(),
        "size:",
        f"{path.stat().st_size / 1024:.2f} KB"
        if path.exists()
        else "N/A"
    )
```

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
while not (PROJECT_ROOT / 'backend').is_dir() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / 'backend').is_dir():
    raise FileNotFoundError('Could not find the HireAI project root. Run this notebook from within the project directory.')
DATASET_PROCESSED = (
    PROJECT_ROOT
    / "ai"
    / "datasets"
    / "processed"
)

V3_TRAIN = DATASET_PROCESSED / "hireai_train_v3.jsonl"
V3_VALIDATION = DATASET_PROCESSED / "hireai_validation_v3.jsonl"
V3_TEST = DATASET_PROCESSED / "hireai_test_v3.jsonl"

print("V3 train:", V3_TRAIN)
print("Exists:", V3_TRAIN.exists())

print("V3 validation:", V3_VALIDATION)
print("Exists:", V3_VALIDATION.exists())

print("V3 test:", V3_TEST)
print("Exists:", V3_TEST.exists())
```

```python
import json

with open(
    V3_TRAIN,
    "r",
    encoding="utf-8"
) as f:

    example = json.loads(
        f.readline()
    )

print("Instruction:")
print(example["instruction"])

print("\nResume:")
print(example["input"])

print("\nTarget:")
print(
    json.dumps(
        example["output"],
        indent=2,
        ensure_ascii=False
    )
)
```

```python
def count_jsonl(path):

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as f:

        return sum(
            1 for _ in f
        )


print(
    "Training:",
    count_jsonl(V3_TRAIN)
)

print(
    "Validation:",
    count_jsonl(V3_VALIDATION)
)

print(
    "Test:",
    count_jsonl(V3_TEST)
)
```

```python
lengths = []

with open(
    V3_TRAIN,
    "r",
    encoding="utf-8"
) as f:

    for line in f:

        example = json.loads(line)

        lengths.append(
            len(example["input"].split())
        )

print("Examples:", len(lengths))
print("Average words:", sum(lengths) / len(lengths))
print("Maximum words:", max(lengths))
print("Minimum words:", min(lengths))
```
