```python
def generate_realistic_experience(
    job_title,
    skills,
    total_years
):
    experience = []

    number_of_jobs = 1

    if total_years >= 4:
        number_of_jobs = random.choice([2, 2, 3])

    remaining_years = total_years
    current_year = 2026

    for job_index in range(number_of_jobs):

        if job_index == number_of_jobs - 1:
            duration = max(1, remaining_years)
        else:
            maximum = max(
                1,
                remaining_years -
                (number_of_jobs - job_index - 1)
            )

            duration = random.randint(
                1,
                min(3, maximum)
            )

        start_year = current_year - duration

        company = random.choice(COMPANIES)

        job_skills = random.sample(
            skills,
            min(4, len(skills))
        )

        descriptions = []

        description_count = random.randint(2, 4)

        for _ in range(description_count):
            template = random.choice(
                BULLET_TEMPLATES
            )

            descriptions.append(
                template.format(
                    skills=", ".join(job_skills)
                )
            )

        experience.append({
            "company": company,
            "job_title": job_title,
            "start_date": f"{start_year}-01",
            "end_date": (
                None
                if job_index == 0
                else f"{current_year}-01"
            ),
            "description": " ".join(
                descriptions
            )
        })

        remaining_years -= duration
        current_year = start_year

        if remaining_years <= 0:
            break

    return experience
```

```python
def format_date_range(start_date, end_date):
    start_year = start_date[:4]

    if end_date is None:
        end = random.choice([
            "Present",
            "Current",
            "Now"
        ])
    else:
        end_year = end_date[:4]

        end = random.choice([
            end_year,
            f"{end_year}-01",
            f"01/{end_year}"
        ])

    start = random.choice([
        start_year,
        start_date,
        f"01/{start_year}"
    ])

    separator = random.choice([
        " - ",
        " – ",
        " to "
    ])

    return f"{start}{separator}{end}"
```

```python
def generate_candidate_v2():

    first_name = random.choice(FIRST_NAMES)
    last_name = random.choice(LAST_NAMES)

    name = f"{first_name} {last_name}"

    email = (
        f"{first_name.lower()}."
        f"{last_name.lower()}@example.com"
    )

    phone = (
        f"+94 7{random.randint(0, 9)} "
        f"{random.randint(100, 999)} "
        f"{random.randint(1000, 9999)}"
    )

    job_title = random.choice(
        list(ROLE_PROFILES.keys())
    )

    available_skills = ROLE_PROFILES[
        job_title
    ]["skills"]

    number_of_skills = random.randint(
        4,
        min(7, len(available_skills))
    )

    skills = random.sample(
        available_skills,
        number_of_skills
    )

    years = random.randint(1, 7)

    experience = generate_realistic_experience(
        job_title,
        skills,
        years
    )

    projects = generate_projects(
        job_title,
        skills
    )

    education = [{
        "degree": random.choice(DEGREES),
        "institution": random.choice(
            UNIVERSITIES
        ),
        "year": random.randint(
            2017,
            2025
        )
    }]

    certifications = []

    if random.random() < 0.7:
        certifications.append(
            random.choice(CERTIFICATIONS)
        )

    if random.random() < 0.25:
        second_cert = random.choice(
            CERTIFICATIONS
        )

        if second_cert not in certifications:
            certifications.append(
                second_cert
            )

    skill_phrases = [
        random.choice(
            SKILL_DESCRIPTIONS.get(
                skill,
                [skill]
            )
        )
        for skill in skills[:3]
    ]

    summary = random.choice([
        (
            f"{job_title} with {years} years "
            f"of experience specializing in "
            f"{', '.join(skill_phrases)}."
        ),

        (
            f"Experienced {job_title} with "
            f"{years} years of professional "
            f"experience in {', '.join(skill_phrases)}."
        ),

        (
            f"Software professional with "
            f"{years} years of experience working "
            f"with {', '.join(skill_phrases)}."
        )
    ])

    return {
        "candidate": {
            "name": name,
            "email": email,
            "phone": phone
        },
        "summary": summary,
        "skills": skills,
        "years_of_experience": years,
        "job_titles": [job_title],
        "experience": experience,
        "education": education,
        "projects": projects,
        "certifications": certifications
    }
```

```python
def generate_projects(job_title, skills):
    number_of_projects = random.randint(1, 3)

    projects = []

    used_names = set()

    for _ in range(number_of_projects):

        available_names = [
            name
            for name in PROJECT_NAMES
            if name not in used_names
        ]

        if not available_names:
            break

        project_name = random.choice(
            available_names
        )

        used_names.add(project_name)

        template = random.choice(
            PROJECT_TEMPLATES[job_title]
        )

        project_skills = random.sample(
            skills,
            min(3, len(skills))
        )

        description = template.format(
            skills=", ".join(project_skills)
        )

        projects.append({
            "name": project_name,
            "description": description
        })

    return projects
```

```python
def resume_layout_1(candidate):
    info = candidate["candidate"]

    text = f"""
{info['name']}
{info['email']} | {info['phone']}

PROFESSIONAL SUMMARY
{candidate['summary']}

TECHNICAL SKILLS
{", ".join(candidate['skills'])}

WORK EXPERIENCE
"""

    for exp in candidate["experience"]:

        date_range = format_date_range(
            exp["start_date"],
            exp["end_date"]
        )

        text += (
            f"\n{exp['job_title']} - "
            f"{exp['company']}\n"
        )

        text += f"{date_range}\n"

        text += (
            f"{exp['description']}\n"
        )

    text += "\nEDUCATION\n"

    for edu in candidate["education"]:
        text += (
            f"{edu['degree']} - "
            f"{edu['institution']} "
            f"({edu['year']})\n"
        )

    text += "\nPROJECTS\n"

    for project in candidate["projects"]:
        text += (
            f"{project['name']}: "
            f"{project['description']}\n"
        )

    if candidate["certifications"]:

        text += "\nCERTIFICATIONS\n"

        for cert in candidate["certifications"]:
            text += f"- {cert}\n"

    return text.strip()
```

```python
def resume_layout_2(candidate):
    info = candidate["candidate"]

    text = (
        f"{info['name']} | "
        f"{info['email']} | "
        f"{info['phone']}\n\n"
    )

    text += (
        f"{candidate['summary']}\n\n"
    )

    text += (
        "Skills: "
        + " | ".join(candidate["skills"])
        + "\n\n"
    )

    text += "Career History\n"

    for exp in candidate["experience"]:
        text += (
            f"\n{exp['job_title']}\n"
            f"{exp['company']}\n"
            f"{format_date_range(exp['start_date'], exp['end_date'])}\n"
        )

        for sentence in re.split(
            r"(?<=\.)\s+",
            exp["description"]
        ):
            text += f"• {sentence}\n"

    text += "\nEducation\n"

    for edu in candidate["education"]:
        text += (
            f"{edu['degree']}, "
            f"{edu['institution']} "
            f"({edu['year']})\n"
        )

    text += "\nSelected Projects\n"

    for project in candidate["projects"]:
        text += (
            f"• {project['name']} - "
            f"{project['description']}\n"
        )

    if candidate["certifications"]:
        text += "\nCertifications\n"

        for cert in candidate["certifications"]:
            text += f"• {cert}\n"

    return text.strip()
```

```python
def resume_layout_3(candidate):
    info = candidate["candidate"]

    text = (
        f"{info['name']}\n"
        f"{info['phone']}\n"
        f"{info['email']}\n\n"
    )

    text += (
        "PROFILE\n"
        f"{candidate['summary']}\n\n"
    )

    text += (
        "CORE COMPETENCIES\n"
        + ", ".join(candidate["skills"])
        + "\n\n"
    )

    text += "EMPLOYMENT HISTORY\n"

    for exp in candidate["experience"]:

        date_range = format_date_range(
            exp["start_date"],
            exp["end_date"]
        )

        text += (
            f"\n{exp['company']} | "
            f"{exp['job_title']}\n"
        )

        text += f"{date_range}\n"

        text += (
            f"- {exp['description']}\n"
        )

    text += "\nACADEMIC QUALIFICATIONS\n"

    for edu in candidate["education"]:
        text += (
            f"{edu['degree']}\n"
            f"{edu['institution']} - "
            f"{edu['year']}\n"
        )

    text += "\nPROJECT EXPERIENCE\n"

    for project in candidate["projects"]:
        text += (
            f"{project['name']}\n"
            f"{project['description']}\n"
        )

    return text.strip()
```

```python
def resume_layout_4(candidate):
    info = candidate["candidate"]

    text = (
        f"{info['name']}\n"
        f"{info['email']} • {info['phone']}\n\n"
    )

    sections = [
        "experience",
        "projects",
        "skills",
        "education",
        "certifications"
    ]

    random.shuffle(sections)

    for section in sections:

        if section == "experience":

            text += "WORK HISTORY\n"

            for exp in candidate["experience"]:
                text += (
                    f"{exp['job_title']} at "
                    f"{exp['company']}\n"
                )

                date_range = format_date_range(
                    exp["start_date"],
                    exp["end_date"]
                )

                text += f"{date_range}\n"

                text += (
                    f"{exp['description']}\n\n"
                )

        elif section == "projects":

            text += "PROJECTS\n"

            for project in candidate["projects"]:
                text += (
                    f"{project['name']}\n"
                    f"{project['description']}\n\n"
                )

        elif section == "skills":

            text += (
                "TECHNOLOGIES\n"
                + " • ".join(candidate["skills"])
                + "\n\n"
            )

        elif section == "education":

            text += "EDUCATION\n"

            for edu in candidate["education"]:
                text += (
                    f"{edu['degree']} - "
                    f"{edu['institution']} "
                    f"{edu['year']}\n\n"
                )

        elif section == "certifications":

            if candidate["certifications"]:
                text += "CERTIFICATIONS\n"

                for cert in candidate["certifications"]:
                    text += f"{cert}\n"

                text += "\n"

    return text.strip()
```

```python
def resume_layout_5(candidate):
    info = candidate["candidate"]

    text = (
        f"{info['name']} | "
        f"{info['email']}\n\n"
        f"{candidate['summary']}\n\n"
    )

    text += "Professional Experience\n"

    for exp in candidate["experience"]:

        text += (
            f"\n{exp['job_title']} "
            f"({exp['company']})\n"
        )

        date_range = format_date_range(
            exp["start_date"],
            exp["end_date"]
        )

        text += f"{date_range}\n"

        sentences = re.split(
            r"(?<=\.)\s+",
            exp["description"]
        )

        for sentence in sentences:
            text += f"* {sentence}\n"

    text += "\nTechnical Skills\n"

    for skill in candidate["skills"]:
        text += f"* {skill}\n"

    text += "\nEducation\n"

    for edu in candidate["education"]:
        text += (
            f"{edu['degree']}\n"
            f"{edu['institution']} "
            f"({edu['year']})\n"
        )

    text += "\nProjects\n"

    for project in candidate["projects"]:
        text += (
            f"{project['name']}\n"
            f"{project['description']}\n"
        )

    return text.strip()
```

```python
RESUME_LAYOUTS = [
    resume_layout_1,
    resume_layout_2,
    resume_layout_3,
    resume_layout_4,
    resume_layout_5
]


def generate_resume_v2(candidate):
    layout = random.choice(
        RESUME_LAYOUTS
    )

    return layout(candidate)
```

```python
def generate_training_example_v2():

    candidate = generate_candidate_v2()

    resume_text = generate_resume_v2(
        candidate
    )

    return {
        "instruction": (
            "Extract structured candidate information "
            "from the resume."
        ),
        "input": resume_text,
        "output": candidate
    }
```

```python
DATASET_SIZE = 1000

dataset_v2 = []

for i in range(DATASET_SIZE):

    dataset_v2.append(
        generate_training_example_v2()
    )

print(
    "Generated:",
    len(dataset_v2)
)
```

```python
seen = set()
unique_dataset_v2 = []

for example in dataset_v2:

    key = example["input"].strip().lower()

    if key not in seen:
        seen.add(key)
        unique_dataset_v2.append(
            example
        )

dataset_v2 = unique_dataset_v2

print(
    "Unique examples:",
    len(dataset_v2)
)

print(
    "Duplicates removed:",
    DATASET_SIZE - len(dataset_v2)
)
```

```python
V2_PATH = (
    DATASET_PROCESSED
    / "hireai_resume_dataset_v2.jsonl"
)

with open(
    V2_PATH,
    "w",
    encoding="utf-8"
) as f:

    for example in dataset_v2:

        f.write(
            json.dumps(
                example,
                ensure_ascii=False
            ) + "\n"
        )

print("Saved:")
print(V2_PATH)
```

```python
random.shuffle(dataset_v2)

total = len(dataset_v2)

train_end = int(total * 0.80)
val_end = int(total * 0.90)

train_v2 = dataset_v2[:train_end]

validation_v2 = dataset_v2[
    train_end:val_end
]

test_v2 = dataset_v2[
    val_end:
]

print("Total:", len(dataset_v2))
print("Train:", len(train_v2))
print("Validation:", len(validation_v2))
print("Test:", len(test_v2))
```

```python
def save_jsonl(records, path):

    with open(
        path,
        "w",
        encoding="utf-8"
    ) as f:

        for record in records:

            f.write(
                json.dumps(
                    record,
                    ensure_ascii=False
                ) + "\n"
            )


TRAIN_V2_PATH = (
    DATASET_PROCESSED
    / "hireai_train_v2.jsonl"
)

VAL_V2_PATH = (
    DATASET_PROCESSED
    / "hireai_validation_v2.jsonl"
)

TEST_V2_PATH = (
    DATASET_PROCESSED
    / "hireai_test_v2.jsonl"
)


save_jsonl(
    train_v2,
    TRAIN_V2_PATH
)

save_jsonl(
    validation_v2,
    VAL_V2_PATH
)

save_jsonl(
    test_v2,
    TEST_V2_PATH
)

print("Datasets saved successfully.")
```
